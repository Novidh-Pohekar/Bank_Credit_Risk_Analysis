```python
from pathlib import Path

import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

candidates = [
    Path.cwd() / "data" / "Banking_Credit_Risk_Dataset.csv",
    Path.cwd().parent / "data" / "Banking_Credit_Risk_Dataset.csv",
]

dataset_path = next((p for p in candidates if p.exists()), candidates[0])
dataset = pd.read_csv(dataset_path)

# Create an empty graph
G = nx.Graph()

# Add edges between Region and Loan Purpose
for _, row in dataset.iterrows():
    G.add_edge(row["Region"], row["Loan_Purpose"])

# Set figure size
plt.figure(figsize=(12, 8))

# Generate layout for node positions
pos = nx.spring_layout(G, seed=42)

# Draw the network graph
nx.draw(
    G,
    pos,
    with_labels=True,
    node_size=3000,
    node_color="skyblue",
    font_size=10,
)

# Add title
plt.title("Regional Loan Purpose Relationship Network")
plt.tight_layout()

# Display the graph
plt.show()
```
